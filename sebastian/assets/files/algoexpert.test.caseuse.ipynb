###### Two Numbers Sum

El objetivo es muy sencillo: buscar dentro del arreglo (`array`) dos números que, al sumarse, den exactamente como resultado el número objetivo (`targetSum`, que en este caso es **10**). Aquí tienes la "lectura" y explicación paso a paso de lo que hace el código:

1. Preparación de los datos

    * **Input:** Tienes una lista de números `[3, 5, -4, 8, 11, 1, -1, 6]` y tu objetivo es llegar a **10**.
    * **`storage = set(num for num in array)`:** Esta línea toma todos los números de tu lista y los guarda en una estructura llamada **conjunto** (`set`).
    * *¿Por qué hace esto?* Porque en programación, buscar si un número existe dentro de un `set` es instantáneo (complejidad temporal de $O(1)$), mientras que buscarlo en una lista normal es mucho más lento porque hay que revisar posición por posición (complejidad de $O(n)$).

2. La lógica principal (El bucle `for`)

    El código empieza a revisar número por número en tu lista original:

    * **`for num in array:`** "Para cada número en mi lista..."
    * **`target = targetSum - num`:** ¡Aquí está la magia matemática! En lugar de sumar todos los números entre sí para ver cuáles dan 10, el código hace una resta. Si tengo el número actual (`num`), ¿cuánto me falta (`target`) para llegar a 10?
    * **`if target in storage and target is not num:`** Aquí se hace dos preguntas:
        1. ¿Ese número que me falta está guardado en mi `set`?
        2. ¿Es un número diferente al que estoy evaluando ahora mismo? (Para evitar usar el mismo número dos veces, como sumar 5 + 5 usando el mismo "5" de la lista).

3. Ejecución paso a paso con tus datos

    Vamos a ver cómo actuaría el código en sus primeras vueltas:

    1. **Primera vuelta:** `num` es **3**.
        * Calcula lo que falta: `10 - 3 = 7`.
        * ¿Está el **7** en nuestro `set`? No. El código avanza.
    2. **Segunda vuelta:** `num` es **5**.
        * Calcula lo que falta: `10 - 5 = 5`.
        * ¿Está el **5** en nuestro `set`? Sí, pero choca con la segunda condición (`target is not num`), porque es el mismo elemento. El código avanza.
    3. **Tercera vuelta:** `num` es **-4**.
        * Calcula lo que falta: `10 - (-4) = 14`.
        * ¿Está el **14** en nuestro `set`? No. El código avanza.
    4. **Cuarta vuelta:** `num` es **8**.
        * Calcula lo que falta: `10 - 8 = 2`.
        * ¿Está el **2** en nuestro `set`? No. El código avanza.
    5. **Quinta vuelta:** `num` es **11**.
        * Calcula lo que falta: `10 - 11 = -1`.
        * ¿Está el **-1** en nuestro `set`? **¡Sí!** Y -1 no es igual a 11.

En este punto, la condición se cumple y el código ejecuta el **`return [num, target]`**, devolviendo la respuesta correcta: **`[11, -1]`**.

---

> **Nota técnica para mejorar tu código:**
> La condición `target is not num` funciona en este caso, pero en Python, la palabra reservada `is` compara la *identidad en memoria* de un objeto, no su *valor matemático*. Para comparar valores y evitar errores extraños con números grandes en el futuro, es una mejor práctica escribirlo utilizando el operador de desigualdad así: **`target != num`**.

```JavaScript
// O(n) Time | O(n) Space
function twoNumberSum(array, targetSum) {
  // Usamos un Set para buscar números en tiempo constante O(1)
  const seen = new Set();
  for (let num of array) {
    // El complemento es el número que nos falta para llegar al targetSum
    const complement = targetSum - num;
    // Si ya vimos el complemento antes, encontramos nuestra pareja
    if (seen.has(complement)) {
      return [num, complement];
    }
    // Si no, marcamos el número actual como visto y pasamos al siguiente
    seen.add(num);
  }
  // Si termina el bucle y no encuentra nada, devuelve un array vacío
  return [];
}
exports.twoNumberSum = twoNumberSum;
```

```Java
import java.util.*;

class Program {
    public static int[] twoNumberSum(int[] array, int targetSum) {
        // Usamos un HashSet para búsqueda eficiente (tiempo O(1))
        Set<Integer> visitedSet = new HashSet<Integer>();
        for (int num : array) {
            int complement = targetSum - num;

            // Si el complemento ya está en el set, encontramos la pareja
            if (visitedSet.contains(complement)) {
                return new int[]{complement, num};
            } else {
                // Si no, agregamos el número actual al set y seguimos
                visitedSet.add(num);
            }
        }
        // Si recorremos todo el array y no encontramos nada, devolvemos un array vacío
        return new int[0];
    }
}
```

```TypeScript
export function twoNumberSum(array: number[], targetSum: number): number[] {
  // Usamos un objeto como mapa para guardar los números que ya vimos
  const numbers: { [n: number]: boolean } = {};
  for (const num of array) {
    const potentialMatch = targetSum - num;
    // ¿Ya vimos el número que nos falta para llegar a la suma?
    if (potentialMatch in numbers) {
      return [potentialMatch, num];
    } else {
      // Si no, guardamos el número actual en nuestro mapa
      numbers[num] = true;
    }
  }
  // Esto solo se ejecuta si terminamos el bucle y no encontramos ninguna pareja
  return [];
}
```
