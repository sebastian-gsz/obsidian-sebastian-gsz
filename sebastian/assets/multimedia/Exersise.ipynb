###### 8 - Números divisibles por pares



- ``def`` : Es la palabra clave en Python para definir una función.
- ``dividir`` : Es el nombre de la función.
- ``*args`` : Es un parámetro especial que permite pasar un número variable de argumentos (en forma de tupla) a la función. Cualquier cantidad de argumentos que se pase a la función será almacenada en args como una tupla.
- Se crea una lista vacía llamada ``resultados`` para almacenar los resultados de las divisiones que se realizarán dentro de la función.
- `if len(args) >= 2:` Aquí se verifica si se han pasado al menos dos argumentos a la función. Si no hay suficientes argumentos, no tiene sentido intentar realizar divisiones.

```python
for i in range(0, len(args), 2):
```

- `range(0, len(args), 2)`: Aquí, `range()` es una función en Python que genera una secuencia de números. Los parámetros que se le pasan son:

   - El primer parámetro es el inicio de la secuencia, que en este caso es 0.
   
   - El segundo parámetro es el final de la secuencia, que es la longitud de la lista `args` que es igual a ``6``.
   
   - El tercer parámetro es el paso, que en este caso es 2. Esto significa que `range` generará números comenzando desde 0 hasta la longitud de `args`, de dos en dos. Por lo tanto, `range(0, len(args), 2)` generará una secuencia de números que representan los índices de la lista `args` empezando desde 0, tomando cada segundo índice.

- `for i in ...`: Esta es la estructura de un bucle `for` en Python que itera sobre cada elemento de la secuencia generada por `range(0, len(args), 2)`. En cada iteración, `i` tomará uno de los valores de la secuencia. La idea detrás del bucle es procesar los elementos de ``args`` de dos en dos :

En cada iteración, el índice ``i`` representa el numerador (``x``).
El índice ``i + 1`` representa el denominador (``y``).
Por ejemplo:

En la primera iteración (``i = 0``):
- ``x = args[0] = 10``
- ``y = args[1] = 2``
- `resultado = 10 / 2 = 5.0`
En la segunda iteración (``i = 2``):
- ``x = args[2] = 8``
- ``y = args[3] = 102``
- ``resultado = 8 / 102 ≈ 0.0784313725490196``
En la tercera iteración (``i = 4``):
- ``x = args[4] = 75``
- ``y = args[5] = 45``
- ``resultado = 75 / 45 ≈ 1.6666666666666667``

- ``resultados.append(resultado)``: En cada iteración con ``append`` añade los resultados a la lista ``resultados`` dentro de la función.

- Existen algunos parámetros como `if y != 0:` si fuera 0 pasaría a la condición ``else``.

Por último se crea una lista `numeros = [10, 2, 8, 102, 75, 45]` y se ejecuta una llamada a la función dividir `resultados = dividir(*numeros)` y se imprime los resultados: ``[5.0, 0.0784313725490196, 1.6666666666666667]``

````python
def pares(*args):
    resultados = []
    if len(args) >= 2:
        for i in range(0, len(args), 2):
            x = args[i]
            y = args[i +1]
            if y != 0:
                resultado = x / y
                resultados.append(resultado)
            else:
                print("Error: división por cero")
    else:
        print("Error: La función 'dividir' necesita al menos dos argumentos")
    return resultados
numeros = [10, 2, 8, 108, 75, 45]
resultado = pares(*numeros)
if resultado:
    average = sum(resultado) / len(resultado)
    print("El promedio (o media aritmética) se calcula dividiendo la suma total de los valores entre la cantidad de valores que hay en el conjunto. \n")
    print(f"Resultados: {resultado}")
    print(f"Promedio de las divisiones: {average}")
else:
    print("No se realizaron divisiones válidas.")
````

###### 9 - Programación estructurada

**- `clientes`**: Es una **lista** que contiene **diccionarios**. Cada diccionario representa a un cliente con su **Nombre**, **Apellidos** y **DNI**.

```python
clientes = [
    {'Nombre': 'Hector', 'Apellidos':'Costa Guzman', 'dni':'11111111A'},
    {'Nombre': 'Juan', 'Apellidos':'González Márquez', 'dni':'22222222B'}
]
```

---

**- `mostrar_cliente(clientes, dni)`**: 

- **Objetivo**: Esta **función** recibe como parámetros la lista de clientes y un **DNI**. Busca en la lista de clientes si existe un cliente con el DNI proporcionado. Si lo encuentra, imprime el **Nombre** y **Apellidos** del cliente. Si no lo encuentra, imprime **'Cliente no encontrado'**.
  
**Detalles**:

- **`for c in clientes:`**: Recorre la lista de clientes uno por uno.
- **`if dni == c['dni']:`**: Compara el **DNI** pasado como parámetro con el **DNI** de cada cliente.
- **`return`**: Si encuentra el cliente, termina la función y no continúa ejecutando el resto del código.

```python
def mostrar_cliente(clientes, dni):
    for c in clientes:
        if dni == c['dni']:
            print('{} {}'.format(c['Nombre'], c['Apellidos']))
            return
    print('Cliente no encontrado')
```

---

**- `borrar_cliente(clientes, dni)`**:

- **Objetivo**: Esta **función** recibe como parámetros la lista de clientes y un **DNI**. Busca un cliente en la lista con ese **DNI** y, si lo encuentra, **elimina** el cliente de la lista y muestra un mensaje indicando que el cliente fue **borrado**. Si no lo encuentra, muestra el mensaje **'Cliente no encontrado'**.

**Detalles**:

- **`enumerate(clientes)`**: Recorre la lista `clientes`, pero en este caso, con **`enumerate`** se obtiene tanto el **índice `i`** como el **valor `c`** (el cliente). El **índice `i`** se usa para eliminar el cliente de la lista.
- **`del(clientes[i])`**: Usa el **índice `i`** para eliminar el cliente en la posición específica dentro de la lista `clientes`.
  
```python
def borrar_cliente(clientes, dni):
    for i, c in enumerate(clientes):  # `i` es el índice, `c` es el diccionario (cliente)
        if dni == c['dni']:  # Compara el DNI proporcionado con el del cliente
            del(clientes[i])  # Elimina el cliente de la lista usando su índice
            print(str(c), "> Cliente Borrado")  # Muestra el cliente eliminado
            return  # Termina la función al encontrar y borrar el cliente
    print('Cliente no encontrado')  # Si no se encuentra el cliente, muestra un mensaje de error
```

---

**Uso de las funciones**:

- Se ejecutan varias funciones para **mostrar clientes** por DNI y **borrar clientes** de la lista. A continuación, mostramos los resultados antes y después de borrar un cliente.

```python
# Uso de las funciones
print("==Listado de clientes==")
print(clientes)  # Muestra todos los clientes antes de hacer cualquier acción.

print("\n==Mostrar clientes por DNI==")
mostrar_cliente(clientes, '11111111A')  # Muestra el cliente con DNI '11111111A'
mostrar_cliente(clientes, '11111111Z')  # Muestra 'Cliente no encontrado' porque no existe el DNI '11111111Z'

print("\n==Borrar clientes por DNI==")
borrar_cliente(clientes, '22222222V')  # Muestra 'Cliente no encontrado' porque no existe el DNI '22222222V'
borrar_cliente(clientes, '22222222B')  # Borra el cliente con DNI '22222222B' y muestra que fue borrado

print("\n==Listado de clientes==")
print(clientes)  # Muestra la lista de clientes después de borrar uno.
```

---

**Explicación del flujo**:

1. **Listado de clientes**: Se muestra la lista completa de los clientes al principio.
2. **Mostrar cliente por DNI**: Se buscan los clientes por su **DNI** usando la función `mostrar_cliente`. En el ejemplo, se busca a `Hector` (que sí existe) y se muestra su nombre, pero al intentar buscar a un cliente con un DNI inexistente (`11111111Z`), la función muestra un mensaje de error.
3. **Borrar cliente por DNI**: Usando `borrar_cliente`, se intenta borrar un cliente. Si el DNI no existe, se muestra un mensaje de error. Si el DNI existe, el cliente es eliminado y se muestra un mensaje indicando que fue borrado.
4. **Listado final de clientes**: Finalmente, se muestra la lista de clientes después de la operación de eliminación.

````python
# Definimos unos cuantos clientes
clientes = [
    {'Nombre': 'Hector', 'Apellidos':'Costa Guzman', 'dni':'11111111A'},
    {'Nombre': 'Juan', 'Apellidos':'González Márquez', 'dni':'22222222B'}
]

# Función que muestra un cliente a partir del DNI
def mostrar_cliente(clientes, dni):
    for c in clientes:
        if dni == c['dni']:
            print('{} {}'.format(c['Nombre'], c['Apellidos']))
            return
    print('Cliente no encontrado')

# Función que borra un cliente a partir del DNI
def borrar_cliente(clientes, dni):
    for i, c in enumerate(clientes):
        if dni == c['dni']:
            del(clientes[i])
            print(str(c), "> Cliente Borrado")
            return
    print('Cliente no encontrado')

# Uso de las funciones
print("==Listado de clientes==")
print(clientes)

print("\n==Mostrar clientes por DNI==")
mostrar_cliente(clientes, '11111111A')
mostrar_cliente(clientes, '11111111Z')

print("\n==Borrar clientes por DNI==")
borrar_cliente(clientes, '22222222V')
borrar_cliente(clientes, '22222222B')

print("\n==Listado de clientes==")
print(clientes)
````

###### 10 - Programación Orientada a Objetos (POO)

**1. Clase `Cliente`**

La clase `Cliente` representa a un cliente de una empresa, con los atributos **DNI**, **nombre** y **apellidos**.

**Método `__init__`**

```python
def __init__(self, dni, nombre, apellidos):
    self.dni = dni
    self.nombre = nombre
    self.apellidos = apellidos
```

- **Propósito**: Es el **constructor** de la clase. Se llama automáticamente cuando creas un nuevo objeto de tipo `Cliente`.

- **Parámetros**:
  - `dni`: El DNI del cliente.
  - `nombre`: El nombre del cliente.
  - `apellidos`: Los apellidos del cliente.
- **Acción**: Asigna los valores pasados a los atributos del objeto:
  - **`self.dni`**: Al atributo `dni` del objeto.
  - **`self.nombre`**: Al atributo `nombre` del objeto.
  - **`self.apellidos`**: Al atributo `apellidos` del objeto.

**Método `__str__`**

```python
def __str__(self):
    return f'{self.nombre} {self.apellidos} (DNI: {self.dni})'
```

- **Propósito**: Proporciona una **representación legible** del objeto cuando se imprime.
- **Acción**: Devuelve una cadena de texto con el formato `Nombre Apellidos (DNI: <dni>)`, que es la forma en que se mostrará el cliente cuando se imprima.

**Método `__repr__`**

```python
def __repr__(self):
    return f"Cliente(dni='{self.dni}', nombre='{self.nombre}', apellidos='{self.apellidos}')"
```

- **Propósito**: Proporciona una **representación técnica** del objeto, más útil para desarrolladores o para depuración.
- **Acción**: Devuelve una cadena con una representación más detallada del objeto, en formato: `Cliente(dni='<dni>', nombre='<nombre>', apellidos='<apellidos>')`. Esta representación es más detallada y muestra todos los atributos de manera más explícita, útil para el debugging.

**2. Clase `Empresa`**

La clase `Empresa` representa una empresa que puede tener una lista de **clientes**.

**Método `__init__`**

```python
def __init__(self, clientes=[]):
    self.clientes = clientes
```

- **Propósito**: Es el **constructor** de la clase `Empresa`. Se usa para crear una instancia de la empresa y asignar una lista de **clientes**.

- **Parámetros**:
  - `clientes`: Una lista de objetos `Cliente` (por defecto, es una lista vacía).
- **Acción**: Asigna la lista de clientes a un atributo llamado `self.clientes`, que es una lista que almacena los clientes asociados a la empresa.

**Método `mostrar_cliente`**

```python
def mostrar_cliente(self, dni=None):
    for c in self.clientes:
        if c.dni == dni:
            print(c)
            return
    print("Cliente no encontrado")
```

- **Propósito**: **Busca un cliente** por su **DNI** en la lista de clientes de la empresa.
- **Parámetros**:
  - `dni`: El DNI del cliente que se quiere mostrar.
- **Acción**:
  - Recorre la lista `self.clientes` buscando un cliente cuyo **DNI** coincida con el proporcionado.
  - Si lo encuentra, imprime el cliente (utilizando **`__str__`** para una representación legible).
  - Si no lo encuentra, imprime **"Cliente no encontrado"**.

**Método `borrar_cliente`**

```python
def borrar_cliente(self, dni=None):
    for i, c in enumerate(self.clientes):
        if c.dni == dni:
            del(self.clientes[i])
            print(str(c), "> Cliente Borrado")
            return
    print("Cliente no encontrado")
```

- **Propósito**: **Borra un cliente** de la lista de clientes de la empresa utilizando su **DNI**.
- **Parámetros**:
  - `dni`: El DNI del cliente que se quiere borrar.
- **Acción**:
  - Recorre la lista `self.clientes` utilizando **`enumerate`** para obtener el índice (`i`) y el cliente (`c`).
  - Si encuentra un cliente con el DNI coincidente, lo elimina de la lista usando `del(self.clientes[i])`.
  - Luego, imprime el cliente borrado (utilizando **`__str__`** para la representación legible).
  - Si no encuentra el cliente, imprime **"Cliente no encontrado"**.

**3. Creación de instancias de las clases**

**Instancia `hector` de la clase `Cliente`**

```python
hector = Cliente(nombre="Hector", apellidos="Costa Guzman", dni="11111111A")
```

- **Propósito**: Crea un objeto `hector` de la clase `Cliente` con los valores proporcionados para `nombre`, `apellidos` y `dni`.
- **Acción**:
  - El **constructor** de la clase `Cliente` asigna los valores `nombre="Hector"`, `apellidos="Costa Guzman"` y `dni="11111111A"` a los atributos correspondientes de `hector`.

**Instancia `juan` de la clase `Cliente`**

```python
juan = Cliente("22222222B", "Juan", "Gonzalez Marquez")
```

- **Propósito**: Crea un objeto `juan` de la clase `Cliente` con los valores proporcionados para `dni`, `nombre` y `apellidos`.
- **Acción**:
  - El **constructor** de la clase `Cliente` asigna los valores `dni="22222222B"`, `nombre="Juan"` y `apellidos="Gonzalez Marquez"` a los atributos correspondientes de `juan`.

**Instancia `empresa` de la clase `Empresa`**

```python
empresa = Empresa(clientes=[hector, juan])
```

- **Propósito**: Crea un objeto `empresa` de la clase `Empresa` con una lista de **clientes** (`hector` y `juan`).
- **Acción**:
  - El **constructor** de la clase `Empresa` asigna la lista `[hector, juan]` al atributo `self.clientes` de `empresa`.

````python
class Cliente:
    def __init__(self, dni, nombre, apellidos):
        self.dni = dni
        self.nombre = nombre
        self.apellidos = apellidos

    def __str__(self):
        return f'{self.nombre} {self.apellidos} (DNI: {self.dni})'

    def __repr__(self):
        # La representación técnica del objeto, útil para los desarrolladores
        return f"Cliente(dni='{self.dni}', nombre='{self.nombre}', apellidos='{self.apellidos}')"

class Empresa:
    def __init__(self, clientes=[]):
        self.clientes = clientes

    def mostrar_cliente(self, dni=None):
        for c in self.clientes:
            if c.dni == dni:
                print(c)
                return
        print("Cliente no encontrado")

    def borrar_cliente(self, dni=None):
        for i, c in enumerate(self.clientes):
            if c.dni == dni:
                del(self.clientes[i])
                print(str(c), "> Cliente Borrado")
                return
        print("Cliente no encontrado")

hector = Cliente(nombre="Hector", apellidos="Costa Guzman", dni="11111111A")
juan = Cliente("22222222B", "Juan", "Gonzalez Marquez")

empresa = Empresa(clientes=[hector, juan])

print("==Listado de clientes==")
print(empresa.clientes)  # Aquí veremos el uso de __repr__

print("\n==Mostrar clientes por DNI==")
empresa.mostrar_cliente("11111111A")
empresa.mostrar_cliente("11111111Z")

print("\n==Borrar clientes por DNI==")
empresa.borrar_cliente("22222222V")
empresa.borrar_cliente("22222222B")

print("\n==Listado de clientes==")
print(empresa.clientes)  # Aquí veremos el uso de __repr__
````
