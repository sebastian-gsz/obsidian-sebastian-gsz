```python
from pathlib import Path
import shutil

# Crear un objeto Path para el directorio actual
directorio_actual = Path(".")

# Crear un directorio llamado "ArchivosTexto" si no existe
directorio_texto = directorio_actual / "ArchivosTexto"
directorio_texto.mkdir(exist_ok=True)

# Crear otro directorio llamado "ArchivosPython" si no existe
directorio_python = directorio_actual / "ArchivosPython"
directorio_python.mkdir(exist_ok=True)

# Crear algunos archivos de ejemplo
archivos_de_ejemplo = [
    "documento1.txt",
    "documento2.txt",
    "script1.py",
    "script2.py",
    "imagen.jpg"
]

for archivo in archivos_de_ejemplo:
    path_archivo = directorio_actual / archivo
    path_archivo.write_text(f"Este es el contenido de {archivo}")

# Listar todos los archivos en el directorio actual
print("Archivos en el directorio actual:")
for archivo in directorio_actual.iterdir():
    if archivo.is_file():
        print(f" - {archivo.name}")

# Filtrar y mover archivos .txt al directorio "ArchivosTexto"
for archivo in directorio_actual.glob("*.txt"):
    print(f"Moviendo {archivo.name} a {directorio_texto}")
    archivo.rename(directorio_texto / archivo.name)

# Filtrar y mover archivos .py al directorio "ArchivosPython"
for archivo in directorio_actual.glob("*.py"):
    print(f"Moviendo {archivo.name} a {directorio_python}")
    archivo.rename(directorio_python / archivo.name)

# Listar archivos en los nuevos directorios
print("\nArchivos en 'ArchivosTexto':")
for archivo in directorio_texto.iterdir():
    if archivo.is_file():
        print(f" - {archivo.name}")

print("\nArchivos en 'ArchivosPython':")
for archivo in directorio_python.iterdir():
    if archivo.is_file():
        print(f" - {archivo.name}")

# Eliminar el directorio "ArchivosTexto" y su contenido
shutil.rmtree(directorio_texto)
print(f"\nDirectorio '{directorio_texto}' eliminado.")

# Eliminar el directorio "ArchivosPython" y su contenido
shutil.rmtree(directorio_python)
print(f"Directorio '{directorio_python}' eliminado.")
```
