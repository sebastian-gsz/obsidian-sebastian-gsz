##### Primera parte: Serialización y deserialización básica

```python
import pickle
from pathlib import Path


# Un objeto Python para serializar
mi_diccionario = {"nombre": "Isabella", "edad": 29, "lenguajes": ["Python", "JavaScript"]}

# Serialización: Guardar el objeto en un archivo
with open("Binario.pickle", "wb") as archivo:
    pickle.dump(mi_diccionario, archivo)

# Deserialización: Cargar el objeto desde el archivo
with open("Binario.pickle", "rb") as archivo:
    diccionario_cargado = pickle.load(archivo)

print(diccionario_cargado)
```

##### Segunda parte: Manejo del diccionario y actualización con nuevos datos

```python
d = dict() # Crea un diccionario vacío
file_name = input("Introduce el nombre del archivo con los datos: ") # Crear o buscar Binario.pickle
path = Path(file_name) # Comprobamos que existe
if path.is_file():
    input_file = open(file_name, 'rb') # Abre el archivo en modo lectura binaria
    d = pickle.load(input_file)
    input_file.close() # Cierra el archivo
else:
    print("El file no existe, creamos diccionario vacio")

document_number = input("Introduce un documento de identidad: ") # Comprueba si está en dict o no.
if document_number in d:
    print("La edad de "+ document_number + " es " + str(d[document_number])) 
else:
    age = input("Documento no existe. Introduce edad: ")
    if age.isnumeric():
        num = int(age)
        d[document_number] = num
        print("Añadimos al diccionario")
        
output_file = open(file_name, 'wb') # Guarda el diccionario en un archivo y lo cierra
pickle.dump(d, output_file)
output_file.close()
```

``import pickle``: Importa la biblioteca pickle, que se usa para la serialización (guardar objetos Python en un archivo) y deserialización (recuperar objetos Python desde un archivo).
``from pathlib import Path``: Importa la clase Path del módulo pathlib, que facilita el trabajo con rutas de archivos de manera orientada a objetos.
``mi_diccionario``: Se crea un diccionario con información personal, incluyendo un nombre, edad, y una lista de lenguajes de programación.

``with open("Binario.pickle", "wb") as archivo:``: Abre un archivo llamado Binario.pickle en modo escritura binaria ("wb"). Si el archivo no existe, lo crea.
``pickle.dump(mi_diccionario, archivo)``: Serializa (convierte en una secuencia de bytes) el diccionario mi_diccionario y lo guarda en el archivo Binario.pickle.

``with open("Binario.pickle", "rb") as archivo:``: Abre el archivo Binario.pickle en modo lectura binaria ("rb").
``diccionario_cargado = pickle.load(archivo)``: Deserializa el contenido del archivo, recuperando el objeto original (mi_diccionario) y lo almacena en diccionario_cargado.

``print(diccionario_cargado)``: Imprime el contenido del diccionario deserializado para verificar que la operación fue exitosa.

``d = dict()``: Crea un diccionario vacío d que será utilizado para almacenar datos adicionales o cargar datos existentes desde un archivo.

``file_name = input("Introduce el nombre del archivo con los datos: ")``: Solicita al usuario que introduzca el nombre de un archivo que contiene datos previamente serializados.
``path = Path(file_name)``: Crea un objeto Path basado en el nombre del archivo proporcionado por el usuario. Este objeto permite manejar de manera sencilla la ruta del archivo y sus propiedades.

``if path.is_file():``: Verifica si el archivo especificado por el usuario existe.
Si el archivo existe:
    ``input_file = open(file_name, 'rb')``: Abre el archivo en modo lectura binaria.
    ``d = pickle.load(input_file)``: Deserializa el contenido del archivo, cargándolo en el diccionario d.
    ``input_file.close()``: Cierra el archivo para liberar los recursos.
Si el archivo no existe:
    ``print("El archivo no existe, creamos diccionario vacío")``: Informa al usuario que el archivo no existe y que se utilizará un diccionario vacío.

``document_number = input("Introduce un documento de identidad: ")``: Solicita al usuario que ingrese un documento de identidad.
``if document_number in d:``: Verifica si el documento de identidad ya existe en el diccionario d.
    Si el documento existe:
        ``print("La edad de "+ document_number + " es " + str(d[document_number]))``: Imprime la edad asociada a ese documento de identidad.
    Si el documento no existe:
        ``age = input("Documento no existe. Introduce edad: ")``: Solicita al usuario que ingrese la edad correspondiente a ese documento.
        ``if age.isnumeric():``: Verifica si la edad ingresada es un número.
            Si es un número:
                ``num = int(age)``: Convierte la edad ingresada de cadena de texto a entero.
                ``d[document_number] = num``: Añade la nueva entrada (documento y edad) al diccionario d.
                ``print("Añadimos al diccionario")``: Informa al usuario que la entrada ha sido añadida al diccionario.

``output_file = open(file_name, 'wb')``: Abre el archivo en modo escritura binaria para guardar el diccionario actualizado.
``pickle.dump(d, output_file)``: Serializa el diccionario d y lo guarda en el archivo.
``output_file.close()``: Cierra el archivo para asegurar que todos los datos se hayan escrito correctamente.
